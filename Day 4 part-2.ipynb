```python
class  Employee():
    name = "Jack"
    role = "CEO"
    def _init_(self,name,role):
        self.name=name
        self.role=role
```

```python
emp2 = Employee()
```

```python
print(emp2.name)
```

```python
print(emp2.role)
```

```python
class BankAccount:
    owner = "Rakhi"
    balance = "100000"
    def _init_(self, owner,balance=0):
        if balance < 0:
            raise ValueError("Initial Balance can not be negative")
        self.owner=owner
        self.balance= balance
```

```python
act1 = BankAccount()
```

```python
print(act1.owner)
```

```python
print(act1.balance)
```
