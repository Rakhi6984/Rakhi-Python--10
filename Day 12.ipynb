# Partition()

```python
text = "Hello World"
print(text.partition("World"))
```

# Startswith()

```python
text = "Python programming"
print(text.startswith("Python"))
```

# Endswith()

```python
text = "Python Programming"
print(text.endswith("Programming"))
```

# Join()

```python
words = ["Python","is","easy"]
text = " ".join(words)
print(text)
```

# Isdigit()

```python
text = "12345"
print(text.isdigit())
```

# Isalpha()

```python
text = "Python"
print(text.isalpha())
```

# Isalnum()

```python
text = "Python123"
print(text.isalnum())
```
