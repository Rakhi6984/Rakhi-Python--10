## String Functions

# Len()

```python
s = "Student"
print(len(s))
```

# Upper()

```python
s= "Hello World"
print(s.upper())
```

# Lower()

```python
s = "Hello World"
print(s.lower())
```

# Concatenation()

```python
s1 = "Hello"
s2 = "World"
print(s1 + " " + s2)
```

```python
s = "Hello"
print(s* 3)
```

# Strip()

```python
s = "Gfg"
print(s.strip())
```

# Replace()

```python
s = "Python is easy"
print(s.replace("easy", "awesome"))
```

# F String()

```python
name = "Alice"
age = 22
print(f"Name:{name}, Age:{age}")
```

# Format String()

```python
name = "Alex"
age = 22
s = "My name is {} and I am {} years old.".format(name, age)
print(s)
```

# TEXT()

```python
text = "Python"
print(text.lower())
```

# Capitalize()

```python
text = "Python programming"
print(text.capitalize())
```

# Title()

```python
text = "python programming language"
print(text. title())
```

# Find()

```python
text = "Python programming"
print(text.find("pro"))
```

# Count()

```python
text = "apple mango apple"
print(text.count("apple"))
```

# Index()

```python
text = "python"
print(text.index("th"))
```

# Split()

```python
text = "apple, banana, mango"
print(text.split(","))
```
