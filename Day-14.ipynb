```python
import cmath 
a = float(input("Enter a: "))
b = float(input("Enter b: "))
c = float(input("Enter c: "))
if a == 0:
    print("This is not a quadratic equation.")
else:
    d = b**2 - 4*a*c
    root1 = (-b + cmath.sqrt(d)) / (2*a)
    root2 = (-b - cmath.sqrt(d)) / (2*a)
    print("Root 1 =", root1)
    print("Root2 =", root2)
```

# Program 5

```python
text = input("Enter a string:")
frequency = {}
for ch in text:
    if ch in frequency:
        frequency[ch] += 1
    else:
        frequency[ch] = 1
print("Character frequency:")
for ch, count in frequency.items():
    print(ch, ":", count)
```

# Program 7

```python
numbers = list(map(int , input("Enter list elements separated by spaces:").split()))
frequency = {}
for num in numbers:
    frequency[num] = frequency.get(num, 0) + 1
unique = []
duplicates = []
for num, count in frequency.items():
    if count == 1:
        unique.append(num)
    else:
        duplicates.append(num)
print("Frequency:", frequency)
print("Unique elements:", unique)
print("Duplicate elements:", duplicates) 
```

# Program 8

```python
def fibonacci(n):
    a, b = 0, 1
    for i in range(n):
        print(a, end= " ")
        a, b = b, a + b
n = int(input("Enter number of terms:"))
if n < 0:
    print("Enter a non-negative number.")
else:
    fibonacci(n)
```

# Program 11

```python
# 1. Single Inheritance 
class Parent:
    def show(self):
        print("Parent class")
class Child(Parent):
    def display(self):
        print("Child Class") 
print("Single Inheritance:")
obj = Child()
obj.show()
obj.display()
# 2. Multiple Inheritance
class Father:
    def father_method(self):
        print("Fathe class")
class Mother:
    def mother_method(self):
        print("Mother class")
class Son(Father, Mother):
    pass
print("\nMultiple Inheritance:")
obj = Son()
obj.father_method()
obj.mother_method()

# 3. Multilevel Inheritance
class A:
    def method_a(self):
        print("Class A")
class B(A):
    def method_b(self):
        print("Class B")
class C(B):
    def method_c(self):
        print("Class C")
print("\nMultilevel Inheritance:")
obj = C()
obj.method_a()
obj.method_b()
obj.method_c()

# 4. Hierarchical Inheritance
class Animal:
    def eat(self):
        print("Animal can eat")
class Dog(Animal):
    def bark(self):
        print("Dog can bark")
class Cat(Animal):
    def meow(self):
        print("Cat can meow")
print("\nHierarchical Inheritance:")
dog = Dog()
dog.eat()
dog.bark()

cat = Cat()
cat.eat()
cat.meow()

# 5. Hybrid Inheritance
class Living:
    def breathe(self):
        print("Living beings breathe")
class Human(Living):
    def human_feature(self):
        print("Human class")
class Worker(Living):
    def work(self):
        print("Worker class")
class Employee(Human, Worker):
    pass 
print("\nHybrid Inheritance:")
obj = Employee()
obj.breathe()
obj.human_feature()
obj.work()
```
