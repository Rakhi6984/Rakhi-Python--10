# Accessing Set Elements using For Loop

```python
fruits = {"apple", "banana", "cherry", "orange"}
print("Iterating through the set:")
for fruit in fruits:
    print(fruit)
```

# Sets Methods

```python
numbers = {10,20,30}
numbers.add(40)
print(sorted(numbers))
```

```python
numbers ={10,20}
numbers.update([30,40,50])
print(sorted(numbers))
```

```python
numbers ={10,20,30}
numbers.remove(20)
print(numbers)
```

```python
numbers ={10,20,30,50}
numbers.discard(50)
print(numbers)
```

```python
numbers ={10,20,30}
x= numbers.pop()
print(x)
print(numbers)
```

```python
numbers = {10,20,30}
numbers.clear()
print(numbers)
```

# Mathematical Operations on Sets

```python
a ={1,2,3,4}
b={3,4,5,6}
a.union(b)
print(a|b)
```

```python
a ={1,2,3,4}
b ={3,4,5,6}
a.intersection(b)
print(a&b)
```

```python
a ={1,2,3,4}
b ={3,4,5,6}
a.difference(b)
print(a-b)
```

```python
a ={1,2,3,4}
b ={3,4,5,6}
a.symmetric_difference(b)
print(a^b)
```

# Membership Operations

```python
numbers ={10,20,30}
print(20 in numbers)
```

```python
numbers ={10,20,30}
print(50 not in numbers)
```

```python
a ={1,2}
b ={1,2,3,4}
print(a.issubset(b))
```

```python
a ={1,2}
b ={1,2,3,4}
print(b.issuperset(a))
```

```python
a ={1,2}
b ={3,4,5,6}
print(a.isdisjoint(b))
```
