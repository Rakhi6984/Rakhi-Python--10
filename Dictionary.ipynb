# Accessing Dictionary Elements

```python
student ={"name":"Rakhi","age":20}
print(student["name"])
print(student["age"])
```

# Dictionary Methods

```python
student ={"name":"Rakhi","age":20}
print(student.keys())
```

```python
student ={"name":"Rakhi","age":20}
print(student.values())
```

```python
student ={"name":"Rakhi","age":20}
print(student.items())
```

```python
student ={"name":"Rakhi","age":20,"course":"bca"}
student.update({"age":21})
print(student)
```

```python
student ={"name":"Rakhi","age":20,"course":"bca"}
print(student.get("name"))
```

```python
student ={"name":"Rakhi","age":20,"course":"bca"}
new_student =student.copy()
print(new_student)
```

```python
student ={"name":"Rakhi","age":20}
student.clear()
print(student)
```

```python
student ={"name":"Rakhi","age":20}
del student["age"]
print(student)
```

```python
student ={"name":"Rakhi","age":20}
age =student.pop("age")
print(student)
```

```python
student ={"name":"Rakhi","age":20}
name =student.popitem()
print(student)
```

# Built-in Methods

```python
student ={"name":"Rakhi","age":20}
print(len(student))
```

```python
student ={"name":"Rakhi","age":20}
print(type(student))
```

```python
student ={"name":"Rakhi","age":20}
print(str(student))
```

```python
student ={"name":"Rakhi","age":20}
print(dict(student))
```

# Traversing Dictionary using Keys

```python
student ={"name":"Rakhi","age":20}
for key in student.keys():
    print(key)
```

# Using Values

```python
student ={"name":"Rakhi","age":20}
for values in student.values():
    print(values)
```

# Membership Operations

```python
student ={"name":"Rakhi","age":20}
print("name" in student)
```

```python
student ={"name":"Rakhi","age":20}
print("marks" not in student)
```
