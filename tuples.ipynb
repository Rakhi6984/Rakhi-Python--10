# Positive Indexing in Tuples

```python
language = "C","C++","Java","Python","Html"
print("First language:",language[0])
print("Third language:",language[2])
```

# Negative Indexing

```python
language = "C","C++","Java","Python","Html"
print("Last language:",language[-1])
print("Second last language:",language[-2])
```

# Slicing

```python
language = "C","C++","Java","Python","Html"
print("First three languages:",language[0:3])
print("Languages index 2 onwards:",language[2:])
print("Languages between index 1 & 4:", language[1:4])
print("Skip Every second language:",language[::2])
```

# Traversing Tuples using For Loop

```python
language = ("C","C++","Java","Python","Html")
print("Traversing using for loop:")
for item in language:
    print(item)
```

# While Loop

```python
language = "C","C++","Java","Python","Html"
print("Traversing using while loop:")
i = 0
while i < len(language):
    print(language[i])
    i += 1
```

# Tuples Packing

```python
student = "Rakhi",20,"BCA"
print(student)
```

# Tuples Unpacking

```python
student = ("Rakhi",20,"BCA")
name,age,course=student
print(name)
print(age)
print(course)
```
