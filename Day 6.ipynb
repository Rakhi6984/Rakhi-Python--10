## VOWELS&CONSONANTS

```python
ch = input("Enter an aplhabet:")
if len(ch) ==1 and ch.isalpha():
    if ch.lower() in ['a', 'e','i','o', 'u']:
        print(ch,"is a Vowel")
    else:
        print(ch,"is a Consonant")
else:
    print("Please enter a single alphabet.")
```
