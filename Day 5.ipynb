##  INHERITANCE

## SINGLE LEVEL INHERITANCE

```python
class vehicle():
    def start(self):
        print("Vehicle Started")
```

```python
Nexon = vehicle()
```

```python
Nexon.start()
```

```python
KIA = vehicle()
```

```python
KIA.start()
```

```python
class car(vehicle):
    def play_music(self):
        print("Playing Music")
```

```python
c1 = car()
```

```python
c1.play_music()
```

```python
c1.start()
```

## MULTIPLE INHERITANCE

```python
class camera:
    def click_photo(self):
        print("Photo Clicked")
```

```python
class phone:
    def call(self):
        print("Calling.....")    
```

```python
class smartphone(camera,phone):
    def phone(self):
        print("I am smartphone")
```

```python
s = smartphone()
```

```python
s.click_photo()
```

```python
s.phone()
```

```python
s.call()
```

## MULTILEVEL INHERITANCE

```python
class animal:
    def eat(self):
        print("I am eating")
```

```python
class dog(animal):
    def bark(self):
        print("Barking")
```

```python
class puppy(dog):
    def weep(self):
        print("Weeping")
```

```python
tommy = puppy()
```

```python
tommy.bark()
```

```python
tommy.weep()
```

```python
class grandfather:
    def house(self):
        print("10 BHK apartment")
```

```python
class father(grandfather):
    def villa(self):
        print("Sea Facing Villa")
```

```python
class child(father):
    def car(self):
        print("BMW")
```

```python
jack = child()
```

```python
jack.villa()
```

```python
jack.car()
```

## HIERARCHICAL INHERITANCE

```python
class vehicle:
    def start(self):
        print("Vehicle Started")
```

```python
class car(vehicle):
    def wheels(self):
        print("4 Wheels")
```

```python
class bike(vehicle):
    def wheels(self):
        print("2 Wheels")
```

```python
c = car()
```

```python
c.start()
```

```python
c.wheels()
```

```python
b = bike()
```

```python
b.start()
```

```python
b.wheels()
```

## HYBRID INHERITANCE

```python
class car(vehicle):
    def wheels(self):
        print("4 wheels")
```
