# FUNCTION

```python
def add(a, b):
    return a + b
result = add(10, 20)
print("Sum =", result)
```

# AREA OF CIRCLE

```python
def area_circle(radius):
    return 3.14 * radius * radius
r = float(input("Enter radius:"))
area = area_circle(r)
print("Area of circle =", area)
```

# POSITIONAL ARGUEMENTS

```python
def student(name, age):
    print("Name:", name)
    print("Age:", age)

student("Riya", 20)
```

# KEYWORD ARGUEMENTS

```python
def student(name, age):
    print("Name:", name)
    print("Age:", age)

student(age = 20, name="Riya")
```

# DEFAULT ARGUEMENTS

```python
def greet(name = "Student"):
    print("Hello", name)

greet()
greet("Riya")    
```

# VARIABLE-LENGTH ARGUEMENTS *ARGS

```python

def add(*args):
    print("Sum=", sum(args))

add(10, 20, 30)
```

# **KWARGS

```python
def display(**kwargs):
    print(kwargs)

display(name = "Riya", age= 20)
```

# TWO TYPES OF ARGUEMENTS

```python
def calculate(a, b=10):
    print("Sum=", a+b)

calculate(5, 20)
calculate(5)
```

# FUNCTIONS TO PERFORM OPERATIONS ON A LIST

```python
def find_max(numbers):
    return max(numbers)

def find_min(numbers):
    return min(numbers)

def find_sum(numbers):
    return sum(numbers)

def find_average(numbers):
    return sum(numbers) / len(numbers)

def search(numbers, element):
    if element in numbers:
        return True
    return False

def frequency(numbers, element):
    return numbers.count(element)

numbers= [10,20,30,20,50]

print("Maximum=", find_max(numbers))
print("Minimum=", find_min(numbers))
print("Sum=", find_sum(numbers))
print("Average=", find_average(numbers))

element = int(input("Enter element to search:"))
if search(numbers, element):
    print("Element found")

else:
    print("Element not found") 

element = int(input("Enter element for frequency:"))
print("Frequency=", frequency(numbers, element))
```

# RECURSIVE TO FIND FACTORIAL OF A GIVEN NUMBER

```python
def factorial(n):
    if n ==0 or n ==1:
        return 1
    else:
        return n * factorial(n -1)
num = int(input("Enter a number:"))
if num < 0:
    print("Factorial is not defined for negative numbers")
else:
    print("Factorial=", factorial(num))
```

# RECURSIVE TO GENERATE THE FIBONACCI SERIES

```python
def fibonacci(n):
    if n <= 1:
        return n
    else:
        return fibonacci(n-1) + fibonacci(n-2)
terms = int(input("Enter number of terms:"))
if terms <= 0:
    print("Enter a positive number")
else:
    print("Fibonacci series:")
    for i in range(terms):
        print(fibonacci(i), end=" ")
```
