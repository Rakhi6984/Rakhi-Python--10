## EXCEPTION HANDLING

```python
try:
    marks = int(input("Enter your Marks:"))
    print("Marks is", marks)
except ValueError:
    print("I think you are entering String But it should be int")
```

```python
def divide (a,b):
    try:
        result = a/b
    except TypeError:
        print("Error: Both Values to be INT String")
    except ZeroDivisionError:
        print("Can not divide by 0")
    else:
        print("Divison Successfull", result)
    finally:
        print("Division Done Successfully")
```

```python
a = int(input("A"))
b = int(input("B"))
```

```python
divide (10,2)
```

```python
divide (10,0)
```

# Class With Attribute

```python
class Employee:
    company = "LW"
    def show_company(self):
        print("Works At:", self.company)

```

```python
emp1 = Employee()
```

```python
emp1.show_company()
```

```python
class product:
    def apply_discount(self, price, percent):
        discounted = price - (price * percent /100)
        return discounted
```

```python
item = product()
```

```python
final_price = item.apply_discount(10000, 25)
print("Final Price After Discount", final_price)
```
